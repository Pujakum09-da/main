```python
# Let's generate a complete data generation code snippet (including issues) in one block for easy copy-paste.

import pandas as pd
import numpy as np

np.random.seed(42)

# Number of employees
n = 200

# Generate sample data
departments = ['Sales', 'HR', 'Tech', 'Operations']
data = {
    'EmployeeID': range(1, n + 1),
    'Department': np.random.choice(departments, n),
    'Tenure': np.random.uniform(0.5, 15, n).round(1),
    'ManagerRating': np.random.randint(2, 6, n),
    'WorkLifeBalance': np.random.randint(2, 6, n),
    'GrowthOpportunities': np.random.randint(1, 6, n),
    'SalarySatisfaction': np.random.randint(2, 6, n),
}

# OverallSatisfaction calculation
data['OverallSatisfaction'] = (
    0.3 * data['ManagerRating'] +
    0.2 * data['WorkLifeBalance'] +
    0.3 * data['GrowthOpportunities'] +
    0.2 * data['SalarySatisfaction'] +
    np.random.normal(0, 0.5, n)
).round().clip(1, 5).astype(int)

# WillStayNextYear
prob_leave = 1 - (data['OverallSatisfaction'] / 5)
data['WillStayNextYear'] = np.where(np.random.rand(n) > prob_leave, 1, 0)

# Create DataFrame
df = pd.DataFrame(data)

# Add missing values
for col in ['ManagerRating', 'WorkLifeBalance', 'GrowthOpportunities', 'SalarySatisfaction']:
    df.loc[df.sample(frac=0.05, random_state=42).index, col] = np.nan

# Add outliers in Tenure
outlier_indices = np.random.choice(df.index, size=3, replace=False)
df.loc[outlier_indices, 'Tenure'] = [40, 50, 60]  # Unrealistically high

# Add duplicate rows
duplicates = df.sample(2, random_state=42)
df = pd.concat([df, duplicates], ignore_index=True)

# Save to CSV
df.to_csv("employee_satisfaction_sample.csv", index=False)

df.head()


```
