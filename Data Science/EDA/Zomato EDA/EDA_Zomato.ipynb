```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
plt.rcParams['figure.figsize']=(12,6)
sns.heatmap(df.isnull(),yticklabels = False,cbar = False,cmap='viridis')
plt.show()
```

```python
df.shape
```

```python
df_country = pd.read_excel('Country-Code.xlsx')
df_country.head()
```

```python
#pd.merge() - for combining the dataset
```

```python
final_df = pd.merge(df,df_country,how='left',on = 'Country Code')
```

```python
final_df.head(2)
```

```python
## to check datatype
```

```python
final_df.dtypes
```

```python
final_df.columns
```

```python
final_df.Country.value_counts()
```

```python
Country_name = final_df.Country.value_counts().index
Country_name
```

```python
Country_val= final_df.Country.value_counts().values
```

```python
##plot pie, Top 3 Countries
plt.pie(Country_val[:3],labels=Country_name[:3],autopct ='%.2f%%')
plt.show()
```

Observations

1. India has most transction
2. USA
3. UK

```python
final_df.columns
```

```python
ratings =final_df.groupby(['Aggregate rating','Rating color','Rating text']).size().reset_index().rename(columns={0:'Rating Columns'})
```

```python
ratings
```

##Observation
1. When Rating is between 4.5 to 4.9 then Excellent
2. When Rating is between 4.0 to 4.4 then Very Good
3. When Rating is between 3.5 to 3.9 then Good
4. When Rating is between 2.5 to 3.4 then Average
5. When Rating is between 1.8 to 2.4 then Poor
6. When Rating is 0.0 then not rated

```python
ratings.head()
```

```python
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=(12,6)
sns.barplot(x='Aggregate rating',y='Rating Columns',data=ratings,palette='viridis')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=(12,6)
sns.barplot(x='Aggregate rating',y='Rating Columns',data=ratings,hue = 'Rating color',palette=['blue','Red','Orange','Yellow','Green','Green'])
plt.show()

##hue - gives different color for each category
```

```python
#Observation
1. Not rated count is very high
2. Maximum people has rated between 2.7 to 3.4
```

```python
sns.countplot(x='Rating color',data=ratings,palette=['blue','Red','Orange','Yellow','Green','Green'])
plt.show()

#here result is based on color count not on value count
```
